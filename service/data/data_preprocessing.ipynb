```python
import pandas as pd

df = pd.read_csv('disease_symptoms.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df.tail()
```

```python
df.rename(columns={df.columns[0]: 'Disease', df.columns[1]: 'Symptoms'}, inplace=True)
df.head()
```

```python
example_data = {
    'records': [
        {
            "Disease":"hypertensive  disease",
            "Symptoms":['pain  chest', 'shortness  of breath', 'dizziness', 'asthenia', 'fall', 'syncope', 'vertigo', 'sweat', 'sweating  increased', 'palpitation', 'nausea', 'angina  pectoris', 'pressure  chest']
        }
    ]
}

print(example_data['records'][0])
```

```python
import json

records = []

for index, row in df.iterrows():
    disease = row['Disease']

    symptoms_str = row['Symptoms']
    symptoms_list = eval(symptoms_str) if isinstance(symptoms_str, str) else symptoms_str
    
    record = {
        "Disease": disease,
        "Symptoms": symptoms_list
    }

    records.append(record)

limit = 20
json_data = {
    'records': records[:limit]
}

print("Sample record from the JSON structure:")
print(json.dumps(json_data['records'][0], indent=4))

# Save the JSON to a file
with open('disease_symptoms.json', 'w') as f:
    json.dump(json_data, f, indent=4)
```
